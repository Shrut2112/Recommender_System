# Popularity Based Recomander System

```python
#merging books dataset with ratings on isbn number comman in both some of the rows might get dropeed
rating_with_name=ratings.merge(books,on="ISBN")
rating_with_name
```

```python
#grouping books by title and counting no. of ratings it received (not what rating how times many rating recieved)
num_rating=rating_with_name.groupby('Book-Title').count()['Book-Rating'].reset_index() # reset index to add index 0 1 2 3 4
num_rating.rename(columns={'Book-Rating':'num-rating'},inplace=True)
num_rating
```

```python
#after grouping books calculating avg-rating from all ratings that book received
avg_rating = rating_with_name.groupby('Book-Title', as_index=False)['Book-Rating'].mean()
avg_rating.rename(columns={'Book-Rating': 'avg-rating'}, inplace=True)
avg_rating
```

```python
#merging dataframe num-rating and avg-rating
popular_df=num_rating.merge(avg_rating,on='Book-Title')
popular_df
```

```python
# finding top 50 movies based on heigest avg-rating from books with more than 250 no. of ratings
popular_df_show=popular_df[popular_df['num-rating']>250].sort_values('avg-rating',ascending=False).head(50)
popular_df_show
```

```python
#merging with book for other info for top 50 books
popular_df_show_display=popular_df_show.merge(books,on='Book-Title').drop_duplicates('Book-Title')[['Book-Title','Book-Author','Year-Of-Publication','Image-URL-M','num-rating','avg-rating']].head(50)
popular_df_show_display
```

# Collaborative Filtering Based Recomander System

```python
rating_with_name
```

Considering rating by only users who rated more than 200 books. And only Books with more than 50 times rating

```python
# finding all the users who have rated more than 200 books
x=rating_with_name.groupby('User-ID').count()['Book-Rating'] >200
padhe_likhe_users=x[x].index
padhe_likhe_users
```

```python
# getting all the data of that users from rating_with_name dataframe
filtered_data=rating_with_name[rating_with_name['User-ID'].isin(padhe_likhe_users)]
filtered_data
```

```python
# finding only those book which are rated more than 50 times
y=filtered_data.groupby('Book-Title').count()['Book-Rating'] >50
famous_books=y[y].index
famous_books
```

```python
# taking all the data from rating_with_name dataframe which has users with more than 200 ratings given and books with more than 50 time rated
final_data=filtered_data[filtered_data['Book-Title'].isin(famous_books)]
final_data
```

```python
# creating a table for collaborative filterring
final_table=final_data.pivot_table(index='Book-Title',columns='User-ID',values='Book-Rating')
final_table.fillna(0,inplace=True)
final_table.head()
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
# using cosine similarity for checking distance between books
similarity_score=cosine_similarity(final_table)
similarity_score
```

```python
# function to recommande books
def recommend(book_name):
  index=np.where(final_table.index==book_name)[0][0] # fetching index
  similar_items=sorted(list(enumerate(similarity_score[index])),key=lambda x:x[1],reverse=True)[1:6]
  data = []
  for i in similar_items:
    item = []
    temp_data = books[books['Book-Title'] == final_table.index[i[0]]]
    item.extend(list(temp_data.drop_duplicates('Book-Title')['Book-Title'].values))
    item.extend(list(temp_data.drop_duplicates('Book-Title')['Book-Author'].values))
    item.extend(list(temp_data.drop_duplicates('Book-Title')['Image-URL-M'].values))
    data.append(item)
  return data
```

```python
recommend("Harry Potter and the Goblet of Fire (Book 4)")
```
